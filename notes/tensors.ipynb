## **Intro**

> Latest version of `Pytorch` is `2.10.0` (as of 2026)

`Pytorch` is a popular open-source Deep Learning Library developed by `Facebook's AI Research lab (FAIR)`. In 2002, `Torch` was developed with `Lua` especially for performing `Tesnsor` based computations. In 2016, `Pytorch` was developed as a Python version of `Torch` and it has been widely used in the field of `Deep Learning` since then.

`Pytorch` provides a flexible and dynamic computational graph. 

**Computational Graph** is a way to represent mathematical computations as a graph, where nodes represent operations and edges represent the flow of data. In `Pytorch`, the computational graph is dynamic, which means that it can be modified on-the-fly during runtime. This allows for more flexibility in building and training neural networks, as you can easily change the architecture or the operations being performed without having to redefine the entire graph.

**ONNX(Open Neural Network Exchange)** is an open format for representing machine learning models. It allows models to be transferred between different frameworks, such as `Pytorch`, `TensorFlow`, and `Caffe2`. This means that you can train a model in one framework and then export it to another framework for inference or deployment.

## **Tensors**

Just like we've other data structures like `arrays`, `lists`, `dataframes`, etc. Each data structure has its own use case. For example, `arrays` are used for numerical computations, `lists` are used for storing collections of items, and `dataframes` are used for tabular data. 

In similar fashion, `tensors` is a specialized multi-dimensional array designed for mathematical and computational tasks.

> Note: Here we are talking about `Tensors` in the context of `Computation`. Actually, `Tensor` is a general term used in mathematics and physics to describe a geometric object that generalizes scalars, vectors, and matrices to higher dimensions and remains consistent under coordinate transformations. 

In `PyTorch`, a `Tensor` is just a multi-dimensional array that can be used to represent data in various forms, such as images, text, and audio.

### **Understanding Tensors with Real World Examples**

**Zero Dimensional Tensor (Scalar)**: A zero-dimensional tensor does not have a direction or orientation. It is simply a single value, like a number. For example, the `temperature` at a specific location can be represented as a zero-dimensional tensor i.e. `30` which represents the temperature in degrees Celsius.

**One Dimensional Tensor (Vector)**: A one-dimensional tensor has a direction and can be represented as a list of values. For example, the `daily temperatures` for a week can be represented as a one-dimensional tensor i.e. `[30, 32, 28, 31, 29, 27, 33]`.

**Two Dimensional Tensor (Matrix)**: A two-dimensional tensor has two directions and can be represented as a table of values. For example, the `pixel values` of a grayscale image can be represented as a two-dimensional tensor, where each value represents the intensity of the pixel i.e. `[[255, 255, 255], [0, 0, 0], [255, 255, 255]]` which represents a simple black and white image.

Imagine 4 students and 3 exam score: 

| Student | Exam 1 | Exam 2 | Exam 3 |
|---------|--------|--------|--------|
| Alice   | 85     | 90     | 95     |
| Bob     | 80     | 85     | 90     |
| Charlie | 75     | 80     | 85     |

Now to represent as a tensor we can: 

```bash
torch.tensor(
  [ # Each inner list represents the scores of a student across the three exams
    [85, 90, 95],   # Each value represents the score of Alice in Exam 1, Exam 2, and Exam 3 respectively
    [80, 85, 90],   
    [75, 80, 85]    
  ]
)
```

Why can't we represent this data as a one-dimensional tensor? Because it has two dimensions: one for the students and one for the exams. Each student has multiple exam scores, so we need a two-dimensional structure to represent this data accurately.

Why can't we represent this data as a three-dimensional tensor? Because we only have two dimensions of data: students and exams. A three-dimensional tensor would imply that we have an additional dimension, such as different classes or subjects, which is not the case here.

**Three Dimensional Tensor**: A three-dimensional tensor has three directions and can be represented as a cube of values. For example, the `RGB pixel values` of a color image can be represented as a three-dimensional tensor, where each value represents the intensity of the red, green, and blue channels i.e. `[[[255, 0, 0], [0, 255, 0], [0, 0, 255]], [[255, 255, 0], [255, 0, 255], [0, 255, 255]], [[255, 255, 255], [128, 128, 128], [0, 0, 0]]]` which represents a simple color image. Whose shape is `(3, 3, 3)` where the first dimension represents the color channels (R, G, B), the second dimension represents the height of the image, and the third dimension represents the width of the image.

Say we've an image of size `3x3` pixels and each pixel has three color channels (Red, Green, Blue). We can represent this image as a three-dimensional tensor where the first dimension represents the color channels, the second dimension represents the height of the image, and the third dimension represents the width of the image.

```bash

torch.tensor(
  [ # color channels
    [ # red channel
    [255, 0, 0], # red channel for the first row of pixels
    [255, 0, 0], # red channel for the second row of pixels
    [255, 0, 0], # red channel for the third row of pixels
  ],
  [ # green channel
    [0, 255, 0], # green channel for the first row of pixels
    [0, 255, 0], # green channel for the second row of pixels
    [0, 255, 0], # green channel for the third row of pixels
  ],
  [ # blue channel
    [0, 0, 255], # blue channel for the first row of pixels
    [0, 0, 255], # blue channel for the second row of pixels
    [0, 0, 255], # blue channel for the third row of pixels
  ]
]
)

```

**Four Dimensional Tensor**: A four-dimensional tensor has four directions and can be represented as a hypercube of values. For example, a `batch of RGB images` can be represented as a four-dimensional tensor, where each value represents the intensity of the red, green, and blue channels for each image in the batch i.e. `[[[[255, 0, 0], [0, 255, 0], [0, 0, 255]], [[255, 255, 0], [255, 0, 255], [0, 255, 255]], [[255, 255, 255], [128, 128, 128], [0, 0, 0]]], [[[255, 0, 0], [0, 255, 0], [0, 0, 255]], [[255, 255, 0], [255, 0, 255], [0, 255, 255]], [[255, 255, 255], [128, 128, 128], [0, 0, 0]]]]` which represents a batch of two color images. Whose shape is `(2, 3, 3, 3)` where the first dimension represents the batch size (number of images), the second dimension represents the color channels (R, G, B), the third dimension represents the height of the image and the fourth dimension represents the width of the image.

<img src="../Notes_Images/4d_tensor.png" alt="4D Tensor" width="600">

**Five Dimensional Tensor**: A five-dimensional tensor has five directions and can be represented as a hypercube of values. For example, a `batch of video frames` can be represented as a five-dimensional tensor, where each value represents the intensity of the red, green, and blue channels for each frame in the video. It represents a batch of two videos with three frames each. Whose shape is `(2 ,3 ,3 ,3 ,3)` where the first dimension represents the batch size (number of videos), the second dimension represents the number of frames in each video (number of images), the third dimension represents the color channels (R ,G ,B), the fourth dimension represents the height of the image and the fifth dimension represents the width of the image.

## Boradcasting

Pytorch aligns tensors on their trailing dimensions. For each dimension, sizes must be equal or one of them must be 1. If the size of a dimension is 1, Pytorch will automatically expand it to match the size of the other tensor in that dimension. This allows for operations to be performed on tensors of different shapes without explicitly reshaping them.

**Broadcasting is only possible when:**

- If the sizes of the tensors are equal, for example, if we've tensors of size `mxn` and `pxq`, then for broadcasting to work, `m = p` and `n = q`. This means that the dimensions of the tensors must be compatible for broadcasting to occur. If `n = p` and `m = q`, then the tensors can be broadcasted together.

- If one of the tensors has a size of 1 in a particular dimension, it can be broadcasted to match the size of the other tensor in that dimension. For example, if we have a tensor of size `mxn` and another tensor of size `1xn`, the second tensor can be broadcasted to match the size of the first tensor in the first dimension.

These work:

```bash
3 and 3 -> compatible, already equal
1 and 3 -> compatible, 1 can stretch to 3
3 and 1 -> compatible, 1 can stretch to 3
```

These don't work:

```bash

2 and 3 -> incompatible
4 and 3 -> incompatible
5 and 2 -> incompatible
```

Create tensor shapes: make tensors with shapes (3,), (3,1), (1,3) and show broadcasting results for addition.

````python
t1 = torch.tensor([1,2,3])

t2 = torch.tensor([
    [0],
    [1],
    [3]
    ])

t3 = torch.tensor([[1,2,3]])

t1.shape, t2.shape, t3.shape
````

````python
broad_add = t2 + t3
broad_add
````

Broadcasting is PyTorch’s way of doing elementwise operations between tensors whose shapes are different but compatible.

Your tensors:

```python
t1 = torch.tensor([1, 2, 3])        # shape: [3]

t2 = torch.tensor([
    [0],
    [1],
    [3]
])                                 # shape: [3, 1]

t3 = torch.tensor([[1, 2, 3]])      # shape: [1, 3]
```

So:

```python
t1.shape  # torch.Size([3])
t2.shape  # torch.Size([3, 1])
t3.shape  # torch.Size([1, 3])
```

When you do:

```python
broad_add = t2 + t3
```

PyTorch compares the shapes from the **rightmost dimension**:

```text
t2:  3 x 1
t3:  1 x 3
```

Dimension by dimension:

```text
last dim:      1 and 3  -> compatible, because one is 1
first dim:     3 and 1  -> compatible, because one is 1
result shape:  3 x 3
```

A dimension of size `1` can be “stretched” virtually to match the other tensor.

So PyTorch treats `t2` like this:

```python
[[0],
 [1],
 [3]]
```

as if it were:

```python
[[0, 0, 0],
 [1, 1, 1],
 [3, 3, 3]]
```

And it treats `t3`:

```python
[[1, 2, 3]]
```

as if it were:

```python
[[1, 2, 3],
 [1, 2, 3],
 [1, 2, 3]]
```

Then it adds element by element:

```text
[[0, 0, 0],      [[1, 2, 3],      [[1, 2, 3],
 [1, 1, 1],   +   [1, 2, 3],   =   [2, 3, 4],
 [3, 3, 3]]       [1, 2, 3]]       [4, 5, 6]]
```

So:

```python
broad_add
```

gives:

```python
tensor([[1, 2, 3],
        [2, 3, 4],
        [4, 5, 6]])
```

Broadcasting does **not usually copy the data** into those larger expanded tensors. It is mostly a virtual expansion: PyTorch behaves as though the smaller tensor were repeated, but internally it can use efficient stride logic.

The general broadcasting rules are:

1. Compare tensor shapes from right to left.
2. Dimensions are compatible if they are equal, or one of them is `1`.
3. If one tensor has fewer dimensions, PyTorch pretends it has leading dimensions of size `1`.
4. The result shape uses the maximum size along each dimension.
5. If any dimension pair is incompatible, PyTorch raises an error.

Example with `t1 + t2`:

```python
t1.shape  # [3]
t2.shape  # [3, 1]
```

PyTorch aligns them from the right:

```text
t2:  3 x 1
t1:      3
```

It treats `t1` as shape:

```text
1 x 3
```

Then:

```text
t2:  3 x 1
t1:  1 x 3
result: 3 x 3
```

So:

```python
t2 + t1
```

also gives:

```python
tensor([[1, 2, 3],
        [2, 3, 4],
        [4, 5, 6]])
```

But this would fail:

```python
a = torch.tensor([[1, 2],
                  [3, 4],
                  [5, 6]])   # shape [3, 2]

b = torch.tensor([1, 2, 3])   # shape [3]

a + b
```

Because shapes align as:

```text
a:  3 x 2
b:      3
```

Compare from the right:

```text
2 and 3 -> incompatible
```

So PyTorch cannot broadcast them.

A helpful mental model:

```text
[3, 1] + [1, 3] -> [3, 3]
```

The `1` dimensions are flexible. They can stretch. Non-`1` dimensions must already match.

Yes. The core thing to see is that PyTorch is doing **element-by-element addition**, so for every output position, it must know exactly which value from `t2` and which value from `t3` to use.

Broadcasting only works when that mapping is unambiguous.

Take the failing case:

```python
t2 = torch.tensor([
    [0, 0],
    [1, 1],
    [3, 3]
])
# shape: [3, 2]

t3 = torch.tensor([[1, 2, 3]])
# shape: [1, 3]
```

Their shapes are:

```text
t2:  3 rows, 2 columns
t3:  1 row,  3 columns
```

Visually:

```text
t2:
[[0, 0],
 [1, 1],
 [3, 3]]

t3:
[[1, 2, 3]]
```

Now PyTorch asks:

Can I stretch `t3`'s rows?

```text
t2 has 3 rows
t3 has 1 row
```

Yes. A size `1` dimension can stretch:

```text
[[1, 2, 3],
 [1, 2, 3],
 [1, 2, 3]]
```

So row-wise, it is fine.

But then PyTorch asks:

Can I match the columns?

```text
t2 has 2 columns
t3 has 3 columns
```

That means PyTorch would need to add:

```text
[0, 0] + [1, 2, 3]
```

But this does not make sense element-by-element:

```text
0 + 1
0 + 2
? + 3
```

There is no third value in `[0, 0]`.

Because `2` and `3` are both **real sizes**, not flexible sizes, PyTorch refuses.

A dimension of size `1` means:

```text
I only have one value here, so it is safe to reuse it.
```

A dimension of size `2` means:

```text
I have two separate values here.
```

A dimension of size `3` means:

```text
I have three separate values here.
```

PyTorch cannot turn two separate values into three separate values automatically, because there is no obvious rule for doing that.

Should it become this?

```text
[0, 0] -> [0, 0, 0]
```

Or this?

```text
[0, 0] -> [0, 0, ?]
```

Or repeat from the start?

```text
[0, 0] -> [0, 0, 0]
```

For `[5, 8]`, should it become:

```text
[5, 8, 5]
```

or:

```text
[5, 8, 8]
```

or something else?

There is no mathematically standard answer. So broadcasting does not allow that.

Now compare that to the working case:

```python
t2 = torch.tensor([
    [0, 0, 0],
    [1, 1, 1],
    [3, 3, 3]
])
# shape: [3, 3]

t3 = torch.tensor([[1, 2, 3]])
# shape: [1, 3]
```

Shapes:

```text
t2: 3 rows, 3 columns
t3: 1 row,  3 columns
```

Now the columns already match:

```text
[0, 0, 0] + [1, 2, 3]
```

That is clear:

```text
0 + 1 = 1
0 + 2 = 2
0 + 3 = 3
```

And because `t3` has only one row, PyTorch can safely reuse that row for every row of `t2`:

```text
t3 virtually becomes:

[[1, 2, 3],
 [1, 2, 3],
 [1, 2, 3]]
```

Then the addition is:

```text
[[0, 0, 0],      [[1, 2, 3],      [[1, 2, 3],
 [1, 1, 1],   +   [1, 2, 3],   =   [2, 3, 4],
 [3, 3, 3]]       [1, 2, 3]]       [4, 5, 6]]
```

The key idea is:

```text
A dimension can only be stretched if it has size 1.
```

Why? Because size `1` means there is only one value along that axis, so repeating it is safe and obvious.

For example:

```python
x = torch.tensor([[10],
                  [20],
                  [30]])
# shape [3, 1]
```

This has 3 rows and 1 column:

```text
[[10],
 [20],
 [30]]
```

If we need 3 columns, each row has only one value, so PyTorch can reuse it:

```text
[[10, 10, 10],
 [20, 20, 20],
 [30, 30, 30]]
```

That is unambiguous.

But:

```python
x = torch.tensor([[10, 20],
                  [30, 40],
                  [50, 60]])
# shape [3, 2]
```

If we need 3 columns, PyTorch cannot safely decide how to turn:

```text
[10, 20]
```

into 3 values.

So it fails.

A very useful mental model:

```text
size 1 = "copy me if needed"
same size = "pair us element by element"
different non-1 sizes = "I don't know how to pair these"
```

So:

```text
[3, 1] + [1, 3] -> works
```

because:

```text
3 and 1 -> stretch 1 to 3
1 and 3 -> stretch 1 to 3
```

But:

```text
[3, 2] + [1, 3] -> fails
```

because:

```text
3 and 1 -> stretch 1 to 3
2 and 3 -> cannot stretch 2 to 3
```

Every dimension must pass the rule.

## **Normalization**

`Normalization` is a technique used to scale the pixel values of an image to a specific range, typically between `0` and `1` or between `-1` and `1`. This is done to improve the performance of machine learning models by ensuring that the input data is on a consistent scale.

**Standardization** is a specific type of normalization where the pixel values are scaled to have a mean of `0` and a standard deviation of `1`. This is done by subtracting the mean from each pixel value and then dividing by the standard deviation. Standardization can help improve the convergence of machine learning models and can also help prevent issues related to different scales of input data.

Convert a uint8 image to float32 normalized to [0,1], move to GPU if available, then back to CPU and convert to NumPy.

````python
# RGB image with 720x1280
img_uint8 = torch.randint(0,255,(3,720,1280),dtype=torch.uint8)

img_uint8
````

````python
import torch
import matplotlib.pyplot as plt

print(img_uint8.shape)
print(img_uint8.dtype)

img_hwc = img_uint8.permute(1, 2, 0)

plt.imshow(img_hwc)
plt.axis("off")
plt.show()
````

````python
# To float
img_uint8 = img_uint8.float()

# Normalize, 0 - 255 -> 0 - 1 i.e. Scale factor is 255 (Min and Max)

img_float = img_uint8 / 255.0

img_float
````

````python
# To GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
img_gpu = img_float.to(device)

img_gpu.device
````

````python
# To CPU
img_numpy = img_gpu.cpu().numpy()

img_numpy.shape

# .cpu()    -> moves tensor back to CPU
# .numpy()  -> converts PyTorch tensor to NumPy array
````
